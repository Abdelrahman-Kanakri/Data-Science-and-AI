# Modelling

```python
x = train1.drop(columns = ["target"])
y = train1["target"]
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size = 0.2, shuffle=True, random_state=42)
```

```python
sc_x = StandardScaler()
sc_y = StandardScaler()

xtrain = sc_x.fit_transform(xtrain)
xtest  = sc_x.transform(xtest)

#ytrain = sc_y.fit_transform(ytrain.values.reshape(-1, 1))
#ytest  = sc_y.transform(ytest.values.reshape(-1, 1))
```

## Neural Network

```python
model = Sequential()
model.add(Dense(8, input_shape=(xtrain.shape[1],), kernel_initializer='normal', activation='relu'))
model.add(Dense(5, activation = "relu", kernel_initializer='normal'))
model.add(Dense(1, activation = 'sigmoid'))

# compiling model
model.compile(optimizer = Adam(learning_rate = 0.01),
              loss = 'binary_crossentropy',
              metrics = ['accuracy'])

model.compile(
    optimizer="adam",
     loss='binary_crossentropy', metrics=['accuracy']
)

early_stopping = EarlyStopping(
    monitor='val_loss',
    patience=5,
    restore_best_weights=True,
    verbose=1
)
```

```python
model.fit(xtrain, ytrain,
          epochs=100,
          validation_split=0.2,
          verbose = 1, 
          callbacks=[early_stopping])
```

```python
preds = model.predict(xtest)
```

```python
preds = (preds > 0.5).astype(int)

accuracy_score(ytest, preds)
```

## RandomForest

```python
rfc = RandomForestClassifier()
```

```python
rfc = RandomForestClassifier(
    n_estimators=1800,         # Number of trees, same as XGB
    max_depth=8,               # Maximum depth of each tree
    max_features=0.3,          # Similar to colsample_bytree (fraction of features)
    bootstrap=True,            # Default behavior, analogous to subsample
    # Note: RandomForest doesn’t have learning rate (eta) or eval_metric params
    random_state=42            # For reproducibility (optional but recommended)
)

rfc.fit(xtrain, ytrain)
```

```python
rfc_pred = rfc.predict(xtest)

rfc_pred = (rfc_pred > 0.5).astype(int)

accuracy_score(ytest, rfc_pred)
```

## Xgboost

```python
xgb = XGBClassifier(
    n_estimators=1800,
    objective="binary:logistic",  # ✅ Binary classification
    booster="gbtree",
    eta=0.01,
    eval_metric="logloss",        # ✅ Binary log loss
    max_depth=8,
    subsample=0.7,
    colsample_bytree=0.3,
    use_label_encoder=False       # ✅ Avoid unnecessary label encoding warning
)

xgb.fit(xtrain, ytrain)
```

```python
xgb_pred = xgb.predict(xtest)

xgb_pred = (xgb_pred > 0.5).astype(int)

accuracy_score(ytest, xgb_pred)
```
